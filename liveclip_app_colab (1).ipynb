Ye notebook aapki zip ka **asli app** (`app.py` + uska UI) Google ke free computer pe chalata hai. Laptop ki zaroorat nahi.

1. Neeche **Cell 1** ka ▶ dabao, phir **Cell 2** ka ▶ dabao.
2. Pehli baar ~2 minute lagte hain (Whisper install hota hai). "Run anyway" aaye to dabao.
3. Niche app khul jayega. **Video file upload karo** → Get preview → Face box / Gameplay box lagao → Make clips.

```python
#@title Cell 1: Files (isse chhedna mat) ▶
ZIP_B64 = ""
print('Files ready.')
```

```python
#@title Cell 2: ▶ App chalao
import os, sys, time, base64, zipfile, io, subprocess, urllib.request
ROOT = "/content/liveclip-main"
if not os.path.exists(ROOT + "/app.py"):
    print("Setup ho raha hai, ~2 min...")
    zipfile.ZipFile(io.BytesIO(base64.b64decode(ZIP_B64))).extractall("/content")
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "faster-whisper", "fastapi", "uvicorn",
                    "python-multipart", "yt-dlp", "pillow", "numpy"], check=True)
if "server" not in globals() or server.poll() is not None:
    server = subprocess.Popen([sys.executable, "app.py", "--no-browser"], cwd=ROOT,
                              env={**os.environ, "PORT": "7860"})
for _ in range(60):
    try:
        urllib.request.urlopen("http://127.0.0.1:7860/api/status", timeout=2); break
    except Exception:
        time.sleep(1)
from google.colab import output
print("App ready. Niche khul raha hai...")
output.serve_kernel_port_as_iframe(7860, height=1400)
```
